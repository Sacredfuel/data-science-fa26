## The Big Idea: Does the Neural Network Earn Its Keep?

Neural networks are powerful, but they are also slower to train, harder to tune and harder to explain than the models from the first half of the course. In Lecture 17, logistic regression on pretrained features beat a frozen ResNet. A network has to **earn** its place.

So every project compares two models on the same data and the same test set:

- **The baseline**: a model from the first half of the course (linear or logistic regression, k-nearest neighbors, decision trees or random forests, ...), trained on simple features: raw columns, flattened pixels, word counts (TF-IDF), etc.
- **The network**: a neural network you train or fine-tune in PyTorch, using at least one technique from the second half of the course (see the tracks below).

**The test:** put the two side by side. Does the network do better? By how much, and at what cost (training time, data, complexity)? Where does each one make mistakes? **A project where the baseline wins is perfectly fine**, as long as you can explain why. That argument is the heart of your report.

**Notes on the Generate track.** Generated images have no "right answer", so you'll evaluate them in three ways:

1. **Look**: grids of samples, interpolations, and the samples you think are best and worst.
2. **Check for copying**: for some generated images, find the nearest training image. Is the model making new images or memorizing old ones?
3. **Ask a classifier**: train a classifier on the real images (transfer learning works well), then check whether it recognizes your generated images as the class you asked for. This also covers the course's requirement of a prediction method.


## Requirements

The purpose of this project is to carry through a data science workflow and put into practice what you have learned in this course in a more open-ended setting than the assignments. Specifically, the project should involve the following steps.

1. Frame a question of your choice that can be addressed by identifying, collecting, and analyzing relevant data.
2. Describe and obtain the data from Kaggle.
3. Perform exploratory data analysis (EDA) and include in your report at least two (but probably many more) data visualizations. For images, this includes a grid of example images per class; for text, examples and length distributions.
4. Describe any data cleaning or transformations that you perform and why they are motivated by your EDA.
5. **Split your data once** into training, validation and test sets (Lecture 16). Make all decisions (model choice, learning rate, number of epochs, ...) with the validation set. Use the test set **once**, at the very end, for both models.
6. Build the **baseline** and the **network**. Explain why each one is appropriate for your question.
7. **Train the network well** (Lecture 16): plot training and validation loss curves, diagnose underfitting or overfitting, and use at least one of dropout, weight decay, data augmentation or early stopping. Describe at least one thing you changed because of what the curves showed.
8. **Compare** the two models on the test set: a table of metrics, model performance plots, and an **error analysis** (a confusion matrix, and a look at examples each model gets wrong).
9. Summarize and interpret your results. Evaluate your approach, discuss limitations, and describe any surprising discoveries and future work.

In order to ensure that you have applied the course materials in sufficient scope, we impose the following additional requirements.

1. The network must be built in **PyTorch** (Hugging Face models count, since they are PyTorch models underneath). You may use pretrained models; say exactly which ones and where they came from.
2. Dataset size:
   - **Tables**: at least six distinct variables (columns) and at least 1,000 rows.
   - **Images or text**: at least 500 labeled examples (or 500 images for the Generate track), and at least 2 classes for classification.
   - Smaller datasets must be approved by me via e-mail.
3. You may read other people's Kaggle notebooks for ideas, but your code must be your own, and you must cite any notebook, tutorial or blog post that influenced your approach.

### Compute

You don't need a powerful laptop:

- **Kaggle Notebooks** and **Google Colab** both offer free GPUs. Kaggle Notebooks can also load any Kaggle dataset directly, without downloading it.
- **Start small.** Get everything working on a subset (a few hundred examples, small images, 1-2 epochs) before training on everything.
- **Images**: resizing to 64x64 or 128x128 (like Lecture 17) makes training far faster. For the Generate track, 32x32 or 64x64 is plenty.
- **Text**: a small pretrained model (e.g. DistilBERT) and a few thousand examples go a long way. Extracting embeddings once and training logistic regression on them (Lecture 17's "features as data") runs fine on a CPU.
- **Save your trained weights** (`torch.save(model.state_dict(), ...)`, Lecture 16) so you don't have to retrain every time you reopen a notebook.

## Tips

- **Start early.** Before you submit your proposal, make sure you can load your dataset and train the baseline on it. That's the quickest way to find out whether the data will work.
- **Build the baseline first.** It's usually a few lines of scikit-learn, it gives you a number to beat, and it often teaches you something about the data.
- **Sanity-check the network before training for real** Make sure it can overfit a single batch, and check that there's no activation function after the output layer.
- **Be suspicious of results that look too good.** If your validation accuracy is 99.9%, look for leakage: duplicate images across splits, a column that gives away the answer, or test data used during training.
- **Look at your data and your mistakes.** A grid of misclassified images or a table of misclassified reviews often explains more than any metric.
- **Accuracy can be misleading** when classes are unbalanced: on a dataset that's 99% "not fraud", predicting "not fraud" every time scores 99%. Report precision and recall, or a confusion matrix, as well.
